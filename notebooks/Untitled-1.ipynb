```python
# Connect DeepEval to Confident AI (set CONFIDENT_API_KEY in .env)
import os
from dotenv import load_dotenv

load_dotenv(override=True)

os.environ.setdefault("DEEPEVAL_PER_ATTEMPT_TIMEOUT_SECONDS_OVERRIDE", "600")

import deepeval
from deepeval.models import OllamaModel

api_key = os.getenv("CONFIDENT_API_KEY")
if not api_key or "paste_your_key" in api_key:
    raise ValueError("Set CONFIDENT_API_KEY in .env")

deepeval.login(api_key=api_key)

judge = OllamaModel(
    model="qwen3:8b",
    base_url="http://localhost:11434",
    temperature=0.0,
)
```

```python
from deepeval.test_case import LLMTestCase
from deepeval.metrics import AnswerRelevancyMetric
from deepeval.models import OllamaModel

if "judge" not in globals():
    judge = OllamaModel(
        model="qwen3:8b",
        base_url="http://localhost:11434",
        temperature=0.0,
    )

answer_relevancy_metric = AnswerRelevancyMetric(model=judge)

test_case = LLMTestCase(
    input="Capital of India",
    actual_output="Delhi",
    retrieval_context=["New Delhi is the capital of India"],
)

score = answer_relevancy_metric.measure(test_case)
print(score)
print(answer_relevancy_metric)
```

```python
from deepeval.test_case import LLMTestCase
from deepeval.metrics import ContextualPrecisionMetric
from deepeval.models import OllamaModel

if "judge" not in globals():
    judge = OllamaModel(
        model="qwen3:8b",
        base_url="http://localhost:11434",
        temperature=0.0,
    )

context_precision_metric = ContextualPrecisionMetric(model=judge)

test_case = LLMTestCase(
    input="Who is the president of the USA in 2024?",
    actual_output="Joe Biden",
    # expected_output="New Delhi is the capital of India",
    retrieval_context=["Donald Trump is the president as of 2022"],
    expected_output="Joe Biden is the current president"
    )



score = context_precision_metric.measure(test_case)
print(context_precision_metric.score)
print(context_precision_metric.success)
print(context_precision_metric.score_breakdown)


```
